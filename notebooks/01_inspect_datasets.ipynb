```python
from datasets import load_dataset
import json, pprint

SUBSETS = ["finqa", "tatqa"]


def truncate(obj, max_str=300, max_list=3):
    """Recursively truncate long strings / lists for readable display."""
    if isinstance(obj, str):
        return obj[:max_str] + " ..." if len(obj) > max_str else obj
    if isinstance(obj, dict):
        return {k: truncate(v, max_str, max_list) for k, v in obj.items()}
    if isinstance(obj, list):
        head = [truncate(i, max_str, max_list) for i in obj[:max_list]]
        return head + ([f"... ({len(obj)-max_list} more)"] if len(obj) > max_list else [])
    return obj


print("Libraries loaded ✓")
```

## 1. FinQA

```python
finqa = load_dataset("rungalileo/ragbench", "finqa", trust_remote_code=True)
print(finqa)
```

```python
# ── Row counts ────────────────────────────────────────────────────────────
print("=== FinQA — Split sizes ===")
for split, ds in finqa.items():
    print(f"  {split:10s}: {len(ds):,} rows")
```

```python
# ── Schema ────────────────────────────────────────────────────────────────
# Use whichever split exists (train > test > validation)
finqa_split = finqa.get("train") or finqa.get("test") or next(iter(finqa.values()))

print("Columns :", finqa_split.column_names)
print()
print("Features:")
pprint.pprint(dict(finqa_split.features))
```

```python
# ── First example ─────────────────────────────────────────────────────────
print("=== FinQA — First example ===")
pprint.pprint(truncate(dict(finqa_split[0])))
```
